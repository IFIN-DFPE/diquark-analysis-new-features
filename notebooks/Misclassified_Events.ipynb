# Misclassified events

```python
from pathlib import Path

results_directory = Path("/data/gmajeri/diquark-analysis/results")
fold_directory = results_directory / "ManyJets/MSuu_8000/wb-ht_jjb-bbt/fold_1"

predictions_path = fold_directory / "gradient_boosting_predictions.npy"
test_set_path = fold_directory / "test_set.parquet"
```

```python
import numpy as np
import pandas as pd

predictions = np.load(predictions_path)
test_df = pd.read_parquet(test_set_path)
```

```python
predictions
```

```python
test_df
```

```python
discriminant = 0.9
predicted_label = predictions > discriminant
```

```python
misclassified = predicted_label & (test_df.target == 0)
misclassified.sum().item()
```

```python
correctly_detected = predicted_label & (test_df.target == 1)
correctly_detected.sum().item()
```

```python
misclassified_sample = test_df[(test_df.jet_multiplicity == 8) & misclassified].iloc[6]

print(misclassified_sample.to_string())
```

```python
correct_sample = test_df[(test_df.jet_multiplicity == 8) & correctly_detected].iloc[0]

print(correct_sample.to_string())
```
