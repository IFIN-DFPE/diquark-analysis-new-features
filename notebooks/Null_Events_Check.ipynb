# Understanding why Pythia generates events with zero jets

```python
import uproot
```

```python
file = uproot.open("/data/gmajeri/diquark-simulations/pythia_with_ptj_10/delphes_output/06_qg_qg.root")

tree = file["Delphes"]
print(*tree.keys(), sep="\n")
```

```python
import awkward as ak

jet_counts = tree.arrays("Jet", library="ak", entry_start=0, entry_stop=1000)["Jet"]

no_jets = jet_counts == 0

print("Number of jets:", jet_counts)
```

```python
print(ak.mean(tree.arrays("Event/Event.Weight", library="ak", entry_start=0, entry_stop=1000)["Event/Event.Weight"]))
```

```python
particle_counts = tree.arrays("Particle", library="ak", entry_start=0, entry_stop=1000)["Particle"][no_jets]
particle_counts
```

```python
particle_pts = tree.arrays("Particle.PT", library="ak", entry_start=0, entry_stop=1000)["Particle.PT"][no_jets]
print("Average particle PTs:", ak.mean(particle_pts, axis=1))
print("Max particle PTs:", ak.max(particle_pts, axis=1))

particle_etas = tree.arrays("Particle.Eta", library="ak", entry_start=0, entry_stop=1000)["Particle.Eta"][no_jets]
print("Average particle etas:", ak.mean(particle_etas, axis=1))
print("Max particle etas:", ak.max(particle_etas, axis=1))
```
