# MadGraph - Matching and Merging + Systematics ON - Event Data Analysis

See also https://pythia.org/latest-manual/CrossSectionsAndWeights.html

````python
import uproot
import awkward as ak
````

## Background generated with matching & merging + systematics OFF

````python
background_events_file = '/data/gmajeri/diquark-simulations/background/low_xqcut_ATLAS_data_136_qcd_100k_MSuu_7.5TeV/Events/run_qcd/tag_1_delphes_events.root'
````

````python
file = uproot.open(background_events_file)
````

````python
file['Delphes'].keys()
````

````python
weights = file['Delphes'].arrays('Event/Event.Weight')['Event/Event.Weight']
weights
````

MadGraph output:
```
  === Results Summary for run: run_qcd tag: tag_1 ===

     Cross-section :   59.39 +- 0.06519 pb
     Nb of events :  100000
     Matched cross-section :   19.95 +- 0.09133 pb
     Nb of events after matching/merging :  33605
     Nb of events after merging :  33605
```

````python
ak.mean(weights)
````

````python
cross_sections = file['Delphes'].arrays('Event/Event.CrossSection')['Event/Event.CrossSection']
cross_sections
````

````python
ak.mean(cross_sections)
````

## Background generated with matching & merging + systematics ON

````python
background_events_file = '/data/gmajeri/diquark-simulations/background/ATLAS_data_136_qcd_50k_MSuu_7.5TeV_ihtmin_3TeV/Events/run_qcd/tag_1_delphes_events.root'
````

````python
file = uproot.open(background_events_file)
````

````python
file['Delphes'].keys()
````

````python
weights = file['Delphes'].arrays('Event/Event.Weight')['Event/Event.Weight']
weights
````

MadGraph output:
```
  === Results Summary for run: run_qcd tag: tag_1 ===

     Cross-section :   2.435 +- 0.003686 pb
     Nb of events :  50000
     Pythia8 merged cross-sections are:
      > Merging scale = 150    : 40353       +/- 2.7     [pb]
      > Merging scale = 225    : 35835       +/- 2.5     [pb]
      > Merging scale = 300    : 31934       +/- 2.4     [pb]
     Notice that because Systematics computation is turned on, the merging did not veto events but modified their weights instead.
     The resulting hepmc/stdhep file should therefore be use with those weights.
```

````python
import awkward as ak

ak.mean(weights)
````

````python
cross_sections = file['Delphes'].arrays('Event/Event.CrossSection')['Event/Event.CrossSection']
cross_sections
````
